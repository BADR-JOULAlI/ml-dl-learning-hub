**Question directrice :** comment passer de douze mesures brutes à une comparaison par catégorie sans perdre les unités ni masquer la dispersion ? Les données sont simulées : elles servent à vérifier le raisonnement, pas à décrire une population.

## Avant de coder

`values` possède la forme `(12,)` : chaque élément est une observation. Votre tableau final aura une ligne par observation. Notez pourquoi `std` et la moyenne ne racontent pas la même chose ; la moyenne mesure le centre, l'écart-type mesure la dispersion. Comparez ensuite les groupes en gardant visible chaque point.

## Tâches

Créez un tableau de 12 mesures ; calculez moyenne et écart-type sans boucle ; placez-le dans un DataFrame et groupez par catégorie ; tracez les observations avec titres et unités.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
rng = np.random.default_rng(4)
values = rng.normal(10, 2, 12)
# Easy: calculate mean and standard deviation with NumPy.
# Medium: build a DataFrame with category labels and aggregate each group.
# Hard: draw a labeled plot and describe one limitation.
```

## Visualisation de départ — voir la dispersion

Ce nuage de points ne calcule pas les réponses. Il montre pourquoi un seul nombre résumant les douze mesures peut cacher des valeurs éloignées. Placez vos statistiques et votre comparaison par groupe dans les cellules que vous ajouterez après ce graphique.

```python
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.scatter(np.arange(len(values)), values, color="tab:blue", s=48)
ax.set(xlabel="Indice de l'observation", ylabel="Mesure (unités arbitraires)", title="Douze mesures simulées")
ax.set_xticks(np.arange(len(values)))
ax.grid(axis="y", alpha=0.2)
plt.show()
```
