## Avant de coder

Avec `y_hat = w*x + b`, la MSE agrège trois erreurs carrées. La différence finie centrée compare la loss à `w+epsilon` et `w-epsilon`. Une approximation numérique proche du gradient analytique est un contrôle utile avant d'entraîner.

## Visualisation de départ — pente locale

La courbe ci-dessous représente la loss lorsque seul `w` varie et que `b=0`. Le minimum visible vous aide à interpréter le signe d'une dérivée. La valeur optimale sur cette coupe n'est pas nécessairement l'optimum lorsque `b` varie aussi.

```python
import matplotlib.pyplot as plt
weight_grid = np.linspace(-1, 4, 120)
loss_grid = [np.mean((candidate * x - y) ** 2) for candidate in weight_grid]
fig, ax = plt.subplots(figsize=(6, 3.5))
ax.plot(weight_grid, loss_grid)
ax.set(xlabel="Pente w (b fixé à 0)", ylabel="MSE", title="Coupe de la surface de perte")
ax.grid(alpha=0.2)
plt.show()
```
