## Visualisation de départ — inspecter les classes

Ce graphique montre une mesure pour chaque classe. Une séparation visible sur une seule feature ne dispense pas du protocole train/test. Ajoutez ensuite une carte des valeurs manquantes et comparez les distributions des partitions.

```python
import matplotlib.pyplot as plt
iris = load_iris(as_frame=True)
fig, ax = plt.subplots(figsize=(6, 3.5))
for class_id, class_name in enumerate(iris.target_names):
    values = iris.data.loc[iris.target == class_id, 'petal length (cm)']
    ax.hist(values, bins=12, alpha=0.45, label=class_name)
ax.set(xlabel="Longueur du pétale (cm)", ylabel="Nombre de fleurs", title="Une variable, trois classes")
ax.legend()
plt.show()
```
